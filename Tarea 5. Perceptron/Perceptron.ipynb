# Perceptrón: compuerta OR

El perceptrón es la red neuronal más simple: una sola neurona. Recibe entradas (x₁, x₂), las multiplica por pesos (w₁, w₂), suma un bias (b) y pasa el resultado z por una función de activación (escalón) que devuelve 0 o 1.

- z = w₁·x₁ + w₂·x₂ + b
- salida = 1 si z ≥ 0, de lo contrario 0

En este ejemplo el perceptrón aprende la compuerta lógica OR a partir de 4 ejemplos. Empieza con pesos y bias en 0 y los corrige cada vez que se equivoca.

```python
import numpy as np

# Datos: compuerta lógica OR
X = np.array([[0, 0], [0, 1], [1, 0], [1, 1]])  # entradas (x1, x2)
y = np.array([0, 1, 1, 1])                       # salida esperada

w = np.zeros(2)   # pesos iniciales (uno por entrada)
b = 0.0           # bias inicial
lr = 0.1          # tasa de aprendizaje (qué tan grandes son los ajustes)

def activacion(z):
    return 1 if z >= 0 else 0   # función escalón

for epoca in range(1, 11):
    errores = 0
    for xi, objetivo in zip(X, y):
        z = np.dot(w, xi) + b          # suma ponderada + bias
        pred = activacion(z)           # salida de la neurona
        error = objetivo - pred        # 0 si acertó, ±1 si falló
        w = w + lr * error * xi        # ajusta los pesos
        b = b + lr * error             # ajusta el bias
        if error != 0:
            errores += 1
    print(f"Época {epoca}: errores={errores}  w={np.round(w,2)}  b={round(b,2)}")
    if errores == 0:
        break

print("Predicciones:", [activacion(np.dot(w, xi) + b) for xi in X])
```

# Resultado

- Los errores bajan de 2 a 0 en 4 épocas: el perceptrón aprendió.
- Las predicciones [0, 1, 1, 1] son la tabla de la verdad del OR.
- Ecuación aprendida: z = 0.1·x₁ + 0.1·x₂ - 0.1. La neurona se activa cuando x₁ + x₂ ≥ 1, es decir, cuando al menos una entrada vale 1.
- El bias es necesario: con la entrada (0,0), w·x vale 0, y sin bias la neurona respondería 1 (incorrecto). El bias de -0.1 lo corrige.

# Cómo aprende

Regla: w = w + lr · error · x, b = b + lr · error, donde error = esperado - predicho.

- Si acierta, el error es 0 y no cambia nada.
- Si dice 0 pero era 1, el error es +1 y suben los pesos de las entradas activas.
- Si dice 1 pero era 0, el error es -1 y bajan.

lr (0.1) controla qué tan grandes son los ajustes. Una época es una pasada completa por los 4 ejemplos.
